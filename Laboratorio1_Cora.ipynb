### ACTIVIDAD B
Determine la cantidad de variables numéricas y no numéricas. Además, indique si existen valores perdidos.

```python
df.info()
```

```python
# Contar var numericas
num_var = df.select_dtypes(include=['number']).shape[1]

# Contar var no numericas
nonum_var = df.select_dtypes(exclude=['number']).shape[1]

# Imprimir las cantidades
print(f"Cantidad de variables numéricas: {num_var}")
print(f"Cantidad de variables no numéricas: {nonum_var}")
```

```python
# Calcular la cantidad de nulos
nulos = df.isnull().sum().sum()

print(f"La cantidad de nulos en el dataframe: {nulos}")
```
